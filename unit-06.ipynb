## Part 3 · Calls counted against truth

**In this part:** you can count the four cells of a confusion table from a truth array and a call array, read one mask through another, sweep a cutoff with a loop, and put scores into bins and take a per-bin mean.

### 3.1 · Four counters from one loop

The setup cell built two arrays of the same length. `truth` holds `True` where an item really is positive. `score` holds a number between 0 and 1 for each item, the kind of number a classifier attaches to a call. A comparison on the whole array, `score >= 0.5`, returns one `True` or `False` per item, and that array is the **call** at a cutoff of 0.5.

Chapter 6 counts the four cells of a confusion table with one loop and four counters. `tp = fp = fn = tn = 0` starts all four at 0 in one line. `zip(call, truth)` walks the two arrays side by side and hands the loop one entry from each on every pass. An `if`/`elif`/`else` then sends each pair to exactly one counter. **Predict the printed call array, then the four counts, before you run the cell.**

```python
call = (score >= 0.5)
print(call)
tp = fp = fn = tn = 0
for called, truly in zip(call, truth):    # one pair per pass: the call and the truth for the same item
    if called and truly:
        tp += 1
    elif called:            # called positive, and truly negative
        fp += 1
    elif truly:             # called negative, and truly positive
        fn += 1
    else:
        tn += 1
print(tp, fp, fn, tn)
```

The call array is `[ True False  True False  True False  True False]`, and the counts are `3 1 1 3`. Four of the eight items are truly positive, and the cutoff caught three of them, so `tp` is 3 and `fn` is 1. Of the four truly negative items, one was called positive, so `fp` is 1 and `tn` is 3. The four counts total 8, the length of the arrays, and that total is the first check to run on any confusion count you write. Each `elif` line carries a condition it does not spell out. `elif called:` runs only when `called and truly` was already `False`, so inside it `truly` must be `False`, which is what the comment on that line says.

### 3.2 · One mask read through another

A mask can go inside the brackets of another mask. `call[truth]` keeps the entries of `call` at the positions where `truth` is `True`, which are the calls made on the truly positive items. `.sum()` on an array of `True` and `False` counts the `True` entries, and `.mean()` gives their share. The chapter writes the other share, the false-alarm rate, with a flipped mask. Comparing a mask with `False`, as in `truth == False`, returns a new mask with every entry flipped. **Predict all four printed lines.**

```python
call = (score >= 0.5)
print(call[truth])
print(call[truth].sum())
print(call[truth].mean())
is_negative = (truth == False)      # flips every entry of truth
print(call[is_negative].mean())
```

The 3 is section 3.1's `tp`, and 0.75 is `tp / (tp + fn)`, the sensitivity. The last line reads `call` at the four truly negative positions, where one call is `True`, so the false-alarm rate is 0.25.

### 3.4 · A bin is two selections in a row

Chapter 6's calibration table collects scores into bins by value, for example 0.4 up to but not including 0.6, and compares the bin's mean score with the share of its items that are truly positive. A bin has a floor and a ceiling, and the chapter applies them one after the other. The first mask keeps everything at or above the floor. The second mask is built from what the first one kept, and keeps what is below the ceiling. `truth` goes through both masks beside `score`, so that the two stay lined up. **Predict the three printed lines for the bin from 0.4 up to 0.6.**

```python
low = 0.4
from_floor_up = (score >= low)            # every item at or above the bin's floor
bin_score = score[from_floor_up]
bin_truth = truth[from_floor_up]
under_ceiling = (bin_score < low + 0.2)   # of those, the ones below its ceiling
bin_score = bin_score[under_ceiling]
bin_truth = bin_truth[under_ceiling]
print(bin_score)
print(len(bin_score), "in the bin")
print("claimed", bin_score.mean(), "  observed", bin_truth.mean())
```

The lines are `[0.55 0.45]`, then `2 in the bin`, then `claimed 0.5   observed 1.0`. Five items pass the floor, and two of those are under the ceiling. `under_ceiling` has five entries, because it was built from `bin_score` after the first selection, and it fits `bin_truth` because `bin_truth` went through the same first selection. The bin's mean score is 0.5, and both of its items are truly positive, so the observed share is 1.0.

The chapter's `calibration_table` is the cell above inside a loop over floors, with a test that skips an empty bin. The cell below is that loop on the setup cell's arrays, printing one line per bin. Every piece of it appeared in this part. **Before running it, predict how many lines come out for a width of 0.5, and what the two numbers on each line are.**

```python
def bin_rows(score, truth, width):
    rows = []
    for low in np.arange(0.0, 1.0, width):
        from_floor_up = (score >= low)
        bin_score = score[from_floor_up]
        bin_truth = truth[from_floor_up]
        under_ceiling = (bin_score < low + width)
        bin_score = bin_score[under_ceiling]
        bin_truth = bin_truth[under_ceiling]
        if len(bin_score) > 0:                   # an empty bin has no mean to report
            rows.append([len(bin_score), bin_truth.mean()])
    return rows

for n_in_bin, observed in bin_rows(score, truth, 0.5):
    print(n_in_bin, "in the bin, observed share", observed)
```

Two lines, `4 in the bin, observed share 0.25` and `4 in the bin, observed share 0.75`. Four scores sit below 0.5, and one of them belongs to a truly positive item. The other four sit at or above 0.5, and three of them do.
